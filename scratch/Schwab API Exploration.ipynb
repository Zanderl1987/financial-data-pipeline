```python
symbol_list = ['VST']

def get_symbol_data(symbol_list,api_key,sleep_time):
  symbol_data_df = pd.DataFrame()
  df_list = []
  
  for symbol in symbol_list:
    end_point_2 = f"https://api.schwabapi.com/marketdata/v1/quotes?symbols={symbol}&fields=quote%2Creference&indicative=false"
    content_2 = requests.get(end_point_2)
    data_2 = content_2.json()
    df1 = pd.DataFrame.from_dict(data_2,orient='index')
    df_list.append(df1)
    #print(df1)
    time.sleep(sleep_time)
    
  symbol_data_df = pd.concat(df_list,axis=0)

  return symbol_data_df
```
